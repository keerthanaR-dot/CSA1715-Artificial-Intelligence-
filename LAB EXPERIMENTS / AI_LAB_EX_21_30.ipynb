3.1 Prolog: Towers of Hanoi

```python
def towers_of_hanoi(n, source, target, auxiliary):
    """
    Solves the Towers of Hanoi puzzle using recursion.
    n: number of disks
    source: name of the source rod
    target: name of the target rod
    auxiliary: name of the auxiliary rod
    """
    if n == 1:
        print(f"Move disk 1 from {source} to {target}")
        return

    # Move top n-1 disks from source to auxiliary using target as auxiliary
    towers_of_hanoi(n - 1, source, auxiliary, target)

    # Move the remaining disk from source to target
    print(f"Move disk {n} from {source} to {target}")

    # Move the n-1 disks from auxiliary to target using source as auxiliary
    towers_of_hanoi(n - 1, auxiliary, target, source)

# Example usage with 3 disks
print("Steps to solve Towers of Hanoi with 3 disks:")
towers_of_hanoi(3, 'A', 'C', 'B')
```

3.2 Prolog: Can a Bird Fly?

```python
def can_bird_fly(species, is_bird=True):
    # Define exceptions of birds that cannot fly
    non_flying_birds = {'penguin', 'ostrich', 'kiwi', 'emu'}

    if not is_bird:
        return False
    if species.lower() in non_flying_birds:
        return False
    return True

# Test cases
test_cases = [
    ('canary', True),
    ('penguin', True),
    ('eagle', True),
    ('ostrich', True),
    ('dog', False)
]

for species, is_bird in test_cases:
    status = "can fly" if can_bird_fly(species, is_bird) else "cannot fly"
    role = "bird" if is_bird else "non-bird"
    print(f"Can a {species} ({role}) fly? -> {status}")
```

3.3 Prolog: Family Tree

```python
# Define parent relationships: parent(Parent, Child)
parents = {
    'John': ['Mary', 'James'],
    'Sophia': ['Mary', 'James'],
    'James': ['David', 'Emma'],
    'Lily': ['David', 'Emma']
}

def get_parents(child):
    """Returns the list of parents for a given child."""
    return [parent for parent, children in parents.items() if child in children]

def get_children(parent):
    """Returns the list of children for a given parent."""
    return parents.get(parent, [])

def are_siblings(person1, person2):
    """Two people are siblings if they share at least one parent and are not the same person."""
    if person1 == person2:
        return False
    p1 = set(get_parents(person1))
    p2 = set(get_parents(person2))
    return len(p1.intersection(p2)) > 0

def get_grandparents(child):
    """Grandparents are the parents of the child's parents."""
    grandparents = []
    for parent in get_parents(child):
        grandparents.extend(get_parents(parent))
    return list(set(grandparents))

# Querying the family tree
print(f"Parents of Mary: {get_parents('Mary')}")
print(f"Children of James: {get_children('James')}")
print(f"Are Mary and James siblings? {are_siblings('Mary', 'James')}")
print(f"Grandparents of David: {get_grandparents('David')}")
```

3.5 Prolog: Monkey and Banana Problem

```python
from collections import deque

# State representation: (monkey_pos, monkey_height, box_pos, has_banana)
# Positions: 'door', 'window', 'middle'
# Height: 'on_floor', 'on_box'

initial_state = ('door', 'on_floor', 'window', False)
goal_state_check = lambda state: state[3] == True

def get_possible_actions(state):
    monkey, height, box, has_banana = state
    actions = []

    if has_banana:
        return actions

    # Action 1: Grasp banana (if monkey is on box and box is at middle/under banana)
    if monkey == 'middle' and height == 'on_box' and box == 'middle':
        actions.append(('grasp', (monkey, height, box, True)))

    # Action 2: Climb box (if monkey and box are at the same position, and monkey is on floor)
    if monkey == box and height == 'on_floor':
        actions.append(('climb', (monkey, 'on_box', box, has_banana)))

    # Action 3: Push box (if monkey and box are at the same position, and monkey is on floor)
    if monkey == box and height == 'on_floor':
        for loc in ['door', 'window', 'middle']:
            if loc != box:
                actions.append((f'push box to {loc}', (loc, height, loc, has_banana)))

    # Action 4: Walk/Go to another position (if monkey is on floor)
    if height == 'on_floor':
        for loc in ['door', 'window', 'middle']:
            if loc != monkey:
                actions.append((f'walk to {loc}', (loc, height, box, has_banana)))

    return actions

def solve_monkey_banana(start):
    # BFS to find the shortest action sequence
    queue = deque([(start, [])])
    visited = {start}

    while queue:
        current_state, path = queue.popleft()

        if goal_state_check(current_state):
            return path

        for action, next_state in get_possible_actions(current_state):
            if next_state not in visited:
                visited.add(next_state)
                queue.append((next_state, path + [action]))

    return None

# Solve the problem
steps = solve_monkey_banana(initial_state)

print("Initial State: Monkey at door, Box at window, Banana hanging in the middle.\n")
print("Solution Steps:")
for idx, step in enumerate(steps, 1):
    print(f"{idx}. {step.capitalize()}")
```

3.6 Prolog: Fruits and Colours Using Backtracking

```python
# Facts database mimicking Prolog facts: fruit(FruitName, Colour)
fruit_colours = [
    ('apple', 'red'),
    ('banana', 'yellow'),
    ('grape', 'purple'),
    ('strawberry', 'red'),
    ('lemon', 'yellow'),
    ('blueberry', 'blue'),
    ('cherry', 'red')
]

def backtrack_search(query_fruit=None, query_colour=None):
    """
    Simulates Prolog's backtracking search.
    Yields matching (fruit, colour) pairs by traversing the facts database.
    """
    for fruit, colour in fruit_colours:
        # Check if the current fact matches the query variables
        match_fruit = (query_fruit is None) or (query_fruit.lower() == fruit)
        match_colour = (query_colour is None) or (query_colour.lower() == colour)

        if match_fruit and match_colour:
            yield (fruit, colour)

# Query 1: What fruits are red? (query_colour='red')
print("--- Query: What fruits are red? ---")
red_fruits = list(backtrack_search(query_colour='red'))
for fruit, colour in red_fruits:
    print(f"- {fruit.capitalize()} is {colour}")

print("\n--- Query: What colour is a banana? ---")
banana_search = list(backtrack_search(query_fruit='banana'))
for fruit, colour in banana_search:
    print(f"- A {fruit} is {colour}")
```

3.7 Prolog: Best First Search

```python
import heapq

# Graph represented as an adjacency list where each edge has a weight (actual distance)
graph = {
    'A': [('B', 3), ('C', 5)],
    'B': [('D', 9), ('E', 8)],
    'C': [('F', 12)],
    'D': [],
    'E': [('G', 5)],
    'F': [('G', 3)],
    'G': []
}

# Heuristic function: Estimated straight-line distance to goal 'G'
heuristics = {
    'A': 15,
    'B': 10,
    'C': 12,
    'D': 8,
    'E': 4,
    'F': 3,
    'G': 0
}

def best_first_search(start, goal):
    # Priority Queue elements: (heuristic_value, current_node, path_taken)
    queue = []
    heapq.heappush(queue, (heuristics[start], start, [start]))

    visited = set()

    while queue:
        # Pop node with the lowest heuristic value (closest to goal)
        h_val, node, path = heapq.heappop(queue)

        if node == goal:
            return path, h_val

        if node not in visited:
            visited.add(node)
            for neighbor, weight in graph.get(node, []):
                if neighbor not in visited:
                    new_path = path + [neighbor]
                    heapq.heappush(queue, (heuristics[neighbor], neighbor, new_path))

    return None, None

# Run Best-First Search from 'A' to 'G'
start_node, goal_node = 'A', 'G'
optimal_path, final_cost = best_first_search(start_node, goal_node)

print(f"Best-First Search path from {start_node} to {goal_node}: {optimal_path}")
if optimal_path:
    print(f"Heuristic estimate at goal: {final_cost}")
```

3.8 Prolog: Medical Diagnosis

```python
# Medical Diagnosis Expert System

# Rules database mapping symptoms to diseases and advice
diagnosis_rules = {
    'flu': {
        'symptoms': {'fever', 'cough', 'body_ache', 'fatigue'},
        'advice': 'Rest, stay hydrated, and consult a doctor if fever persists.'
    },
    'common_cold': {
        'symptoms': {'cough', 'runny_nose', 'sneezing', 'sore_throat'},
        'advice': 'Keep warm, drink warm fluids, and take vitamin C.'
    },
    'food_poisoning': {
        'symptoms': {'nausea', 'vomiting', 'diarrhea', 'stomach_cramps'},
        'advice': 'Stay hydrated with electrolytes, eat bland food, and rest.'
    },
    'allergy': {
        'symptoms': {'sneezing', 'watery_eyes', 'itchy_skin', 'runny_nose'},
        'advice': 'Avoid allergens, and use antihistamines if advised by a pharmacist.'
    }
}

def diagnose_patient(patient_symptoms):
    """
    Infers potential diseases based on matching the patient's symptoms with the rules database.
    """
    patient_symptoms_set = {s.strip().lower() for s in patient_symptoms}
    diagnoses = []

    for disease, data in diagnosis_rules.items():
        # Count how many symptoms match
        matching = patient_symptoms_set.intersection(data['symptoms'])
        match_ratio = len(matching) / len(data['symptoms'])

        # If at least 50% of the disease symptoms match, we flag it as a possibility
        if match_ratio >= 0.5:
            diagnoses.append({
                'disease': disease.replace('_', ' ').capitalize(),
                'confidence': f"{match_ratio * 100:.1f}%",
                'advice': data['advice']
            })

    return diagnoses

# Scenario 1: Patient complains of fever, fatigue, and cough
symptoms_1 = ['fever', 'fatigue', 'cough']
print(f"Patient Symptoms: {symptoms_1}")
results_1 = diagnose_patient(symptoms_1)
for r in results_1:
    print(f"\nPotential Diagnosis: {r['disease']} (Match: {r['confidence']})")
    print(f"Advice: {r['advice']}")

print("\n" + "="*40 + "\n")

# Scenario 2: Patient complains of sneezing and runny nose
symptoms_2 = ['sneezing', 'runny_nose']
print(f"Patient Symptoms: {symptoms_2}")
results_2 = diagnose_patient(symptoms_2)
for r in results_2:
    print(f"\nPotential Diagnosis: {r['disease']} (Match: {r['confidence']})")
    print(f"Advice: {r['advice']}")
```

```python
def interactive_diagnosis():
    print("=== Interactive Medical Diagnosis Expert System ===")
    print("Available symptoms in our database: fever, cough, body_ache, fatigue, runny_nose, sneezing, sore_throat, nausea, vomiting, diarrhea, stomach_cramps, watery_eyes, itchy_skin\n")

    user_input = input("Please enter your symptoms separated by commas (e.g., fever, cough): ")

    # Split the input by commas and clean up whitespaces
    symptoms = [s.strip() for s in user_input.split(',') if s.strip()]

    if not symptoms:
        print("No symptoms entered. Exiting.")
        return

    print(f"\nAnalyzing entered symptoms: {symptoms}")
    results = diagnose_patient(symptoms)

    if not results:
        print("\nNo matching diseases found in our expert system based on your symptoms.")
    else:
        for r in results:
            print(f"\nPotential Diagnosis: {r['disease']} (Match Confidence: {r['confidence']})")
            print(f"Advice: {r['advice']}")

# Run the interactive diagnostic tool
interactive_diagnosis()
```

3.9 Prolog: Forward Chaining

```python
class Rule:
    def __init__(self, premises, conclusion):
        self.premises = set(premises)
        self.conclusion = conclusion

    def __repr__(self):
        return f"IF {', '.join(self.premises)} THEN {self.conclusion}"

# 1. Define the knowledge base of rules
rules = [
    Rule(['A', 'B'], 'C'),  # IF A and B are true, THEN C is true
    Rule(['C'], 'D'),       # IF C is true, THEN D is true
    Rule(['D', 'E'], 'F'),  # IF D and E are true, THEN F is true
    Rule(['G'], 'A')        # IF G is true, THEN A is true
]

# 2. Define our initial facts (known truths)
initial_facts = {'A', 'B', 'E'}

def forward_chain(rules, facts):
    print(f"Initial Facts: {sorted(list(facts))}\n")
    known_facts = set(facts)
    inferred_new_fact = True
    step = 1

    while inferred_new_fact:
        inferred_new_fact = False
        for rule in rules:
            # If the rule's conclusion is already known, skip it
            if rule.conclusion in known_facts:
                continue

            # If all premises are met by known facts, infer the conclusion
            if rule.premises.issubset(known_facts):
                known_facts.add(rule.conclusion)
                print(f"Step {step}: Rule [{rule}] matched. Inferring and adding '{rule.conclusion}' to facts.")
                print(f"Current Facts: {sorted(list(known_facts))}\n")
                inferred_new_fact = True
                step += 1
                break  # Break and restart the rule evaluation loop with updated facts

    return known_facts

# Run the forward chaining engine
final_facts = forward_chain(rules, initial_facts)
print(f"Final derived facts: {sorted(list(final_facts))}")
```

```python
class Rule:
    def __init__(self, premises, conclusion):
        self.premises = premises  # list of premises needed
        self.conclusion = conclusion  # the single proposition concluded

    def __repr__(self):
        return f"IF {', '.join(self.premises)} THEN {self.conclusion}"

# 1. Define the knowledge base of rules
rules = [
    Rule(['A', 'B'], 'C'),  # To prove C, we need to prove A and B
    Rule(['C'], 'D'),       # To prove D, we need to prove C
    Rule(['D', 'E'], 'F'),  # To prove F, we need to prove D and E
    Rule(['G'], 'A')        # To prove A, we need to prove G
]

# 2. Define our initial facts (known truths)
facts = {'A', 'B', 'E'}

def backward_chain(goal, rules, facts, visited_goals=None):
    if visited_goals is None:
        visited_goals = set()

    print(f"Goal: Prove '{goal}'")

    # Case 1: The goal is already a known fact
    if goal in facts:
        print(f"-> '{goal}' is a known initial fact. Proved!")
        return True

    # Prevent infinite loops in recursive goals
    if goal in visited_goals:
        print(f"-> Cycle detected for goal '{goal}'. Cannot prove.")
        return False

    visited_goals.add(goal)

    # Case 2: Find all rules that have the goal as their conclusion
    candidate_rules = [r for r in rules if r.conclusion == goal]

    if not candidate_rules:
        print(f"-> No rules found that conclude '{goal}'. Cannot prove.")
        return False

    # Try to prove premises of at least one candidate rule
    for rule in candidate_rules:
        print(f"-> Found rule: [{rule}]. Attempting to prove premises: {rule.premises}")
        all_premises_proved = True

        for premise in rule.premises:
            if not backward_chain(premise, rules, facts, visited_goals.copy()):
                all_premises_proved = False
                print(f"-> Premise '{premise}' failed for rule [{rule}].")
                break  # Try next rule

        if all_premises_proved:
            print(f"-> Successfully proved all premises of [{rule}]. Concluding '{goal}'!")
            facts.add(goal)  # Memoize this proved goal as a new fact
            return True

    return False

# Run the backward chaining solver to verify if we can prove the goal 'F'
target_goal = 'F'
print(f"=== Starting Backward Chaining for Goal '{target_goal}' ===\n")
success = backward_chain(target_goal, rules, facts)
print(f"\nFinal Result: Was goal '{target_goal}' proved? -> {success}")
```
