2.1 Map Colouring (CSP)

```python
colors = ["Red", "Green", "Blue"]
graph = {"A":["B","C"], "B":["A","C","D"],
         "C":["A","B","D"], "D":["B","C"]}
result = {}

def solve(node):
    if node == len(graph):
        return True
    region = list(graph)[node]
    for color in colors:
        if all(result.get(n) != color for n in graph[region]):
            result[region] = color
            if solve(node+1):
                return True
            del result[region]
    return False

solve(0)
print(result)
```

2.2 Tic Tac Toe Game

```python
b = [" "] * 9

def show():
    for i in range(0, 9, 3):
        print("|".join(b[i:i+3]))
        print("-" * 5)

for turn in range(9):
    show()
    p = "X" if turn % 2 == 0 else "O"
    pos = int(input(p + " enter position (1-9): ")) - 1
    if 0 <= pos < 9 and b[pos] == " ":
        b[pos] = p
        if any(all(b[i] == p for i in line) for line in
               [(0,1,2),(3,4,5),(6,7,8),(0,3,6),
                (1,4,7),(2,5,8),(0,4,8),(2,4,6)]):
            show()
            print(p, "wins!")
            break
else:
    show()
    print("Draw!")
```

2.3 Minimax Algorithm for Gaming

```python
def minimax(depth, maximizing, scores):
    if depth == 2:
        return scores.pop(0)
    values = [minimax(depth+1, not maximizing, scores)
              for _ in range(2)]
    return max(values) if maximizing else min(values)

print("Best score:", minimax(0, True, [3, 5, 2, 9]))
```

2.4 Alpha-Beta Pruning Algorithm

```python
def alphabeta(depth, node, alpha, beta, values):
    if depth == 2:
        return values.pop(0)
    if node:
        best = -float("inf")
        for _ in range(2):
            best = max(best, alphabeta(depth+1, False, alpha, beta, values))
            alpha = max(alpha, best)
            if beta <= alpha:
                break
        return best
    else:
        best = float("inf")
        for _ in range(2):
            best = min(best, alphabeta(depth+1, True, alpha, beta, values))
            beta = min(beta, best)
            if beta <= alpha:
                break
        return best

print("Best score:", alphabeta(0, True, -float("inf"),
                               float("inf"), [3, 5, 2, 9]))
```

2.7 Prolog: Sum of Integers from 1 to n

```python
# Prolog equivalent code (in comments):
# sum(0, 0).
# sum(N, Sum) :-
#     N > 0,
#     N1 is N - 1,
#     sum(N1, Sum1),
#     Sum is N + Sum1.

def prolog_sum(n):
    # Base case: sum(0, 0)
    if n == 0:
        return 0
    # Recursive step
    elif n > 0:
        return n + prolog_sum(n - 1)

# Test the function
n = 10
print(f"Sum of integers from 1 to {n} is: {prolog_sum(n)}")
```

2.8 Prolog: Database with Name and DOB

```python
# Prolog code (in comments):
# % Facts
# dob(john, '1995-05-15').
# dob(alice, '1998-10-22').
# dob(bob, '1993-02-12').
#
# % Rule to query DOB
# get_dob(Name, DOB) :- dob(Name, DOB).

# Python simulation of a Prolog Database
database = {
    "john": "1995-05-15",
    "alice": "1998-10-22",
    "bob": "1993-02-12"
}

def query_dob(name):
    if name in database:
        return f"{name}'s Date of Birth is {database[name]}."
    else:
        return f"No record found for {name}."

# Query the database
print(query_dob("alice"))
print(query_dob("john"))
print(query_dob("charlie"))
```

2.9 Prolog: Student - Teacher - Subject Code

```python
# Prolog equivalent code (in comments):
# % Facts: studies(Student, Subject)
# studies(john, math).
# studies(alice, science).
# studies(bob, history).
#
# % Facts: teaches(Teacher, Subject)
# teaches(mr_smith, math).
# teaches(ms_davis, science).
# teaches(mr_jones, history).
#
# % Rule: Who teaches which student?
# teacher_of(Teacher, Student) :-
#     studies(Student, Subject),
#     teaches(Teacher, Subject).

# Python relational database simulation
studies = [
    ("john", "math"),
    ("alice", "science"),
    ("bob", "history")
]

teaches = [
    ("mr_smith", "math"),
    ("ms_davis", "science"),
    ("mr_jones", "history")
]

def find_teacher_of(student_name):
    # Find subjects the student studies
    student_subjects = [subject for student, subject in studies if student == student_name]

    results = []
    for subject in student_subjects:
        # Find teachers for those subjects
        teachers_for_subject = [teacher for teacher, subj in teaches if subj == subject]
        for teacher in teachers_for_subject:
            results.append((teacher, subject))
    return results

# Query the relationships
student_to_query = "alice"
matches = find_teacher_of(student_to_query)

for teacher, subject in matches:
    print(f"{teacher} teaches {student_to_query} {subject}.")
```

2.10 Prolog: Planets Database

```python
# Prolog equivalent code (in comments):
# % Facts: planet(Name, Type, OrderFromSun)
# planet(mercury, rocky, 1).
# planet(venus, rocky, 2).
# planet(earth, rocky, 3).
# planet(mars, rocky, 4).
# planet(jupiter, gas_giant, 5).
# planet(saturn, gas_giant, 6).
# planet(uranus, ice_giant, 7).
# planet(neptune, ice_giant, 8).
#
# % Rule: Is the planet a giant?
# is_giant(Name) :- planet(Name, gas_giant, _).
# is_giant(Name) :- planet(Name, ice_giant, _).

# Python simulation of a Prolog Database for Planets
planets = [
    {"name": "mercury", "type": "rocky", "order": 1},
    {"name": "venus", "type": "rocky", "order": 2},
    {"name": "earth", "type": "rocky", "order": 3},
    {"name": "mars", "type": "rocky", "order": 4},
    {"name": "jupiter", "type": "gas_giant", "order": 5},
    {"name": "saturn", "type": "gas_giant", "order": 6},
    {"name": "uranus", "type": "ice_giant", "order": 7},
    {"name": "neptune", "type": "ice_giant", "order": 8}
]

def is_giant(planet_name):
    for p in planets:
        if p["name"] == planet_name:
            return p["type"] in ["gas_giant", "ice_giant"]
    return False

def get_planets_by_type(planet_type):
    return [p["name"] for p in planets if p["type"] == planet_type]

# Queries
print("Is earth a giant planet?:", is_giant("earth"))
print("Is jupiter a giant planet?:", is_giant("jupiter"))
print("Rocky planets in our database:", get_planets_by_type("rocky"))
```
