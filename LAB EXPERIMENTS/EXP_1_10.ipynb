1.1 8-Puzzle Problem

```python
from collections import deque

start = (1,2,3,4,0,6,7,5,8)
goal = (1,2,3,4,5,6,7,8,0)
q = deque([(start, [])])
seen = {start}

while q:
    s, path = q.popleft()
    if s == goal:
        print("Solution:", path + [s])
        break
    z = s.index(0)
    for n in [z-3, z+3, z-1, z+1]:
        if 0 <= n < 9 and not (abs(n-z)==1 and z//3 != n//3):
            t = list(s)
            t[z], t[n] = t[n], t[z]
            t = tuple(t)
            if t not in seen:
                seen.add(t)
                q.append((t, path+[s]))
```

1.2 8-Queen Problem


```python
def solve(board, row=0):
    if row == 8:
        return True
    for col in range(8):
        if all(board[i] != col and abs(board[i]-col) != row-i
               for i in range(row)):
            board[row] = col
            if solve(board, row+1):
                return True
    return False

b = [-1]*8
solve(b)
for c in b:
    print("."*c + "Q" + "."*(7-c))
```

1.3 Water Jug Problem

```python
from collections import deque

q = deque([(0, 0, [])])
seen = set()

while q:
    a, b, path = q.popleft()
    if (a, b) in seen:
        continue
    seen.add((a, b))
    path = path + [(a, b)]
    if a == 2 or b == 2:
        print(path)
        break
    q.extend([
        (4,b,path), (a,3,path), (0,b,path), (a,0,path),
        (a-min(a,3-b), b+min(a,3-b),path),
        (a+min(b,4-a), b-min(b,4-a),path)
    ])
```

1.4 Crypt-Arithmetic Problem

```python
from itertools import permutations

for p in permutations(range(10), 8):
    S,E,N,D,M,O,R,Y = p
    if S and M:
        send = 1000*S+100*E+10*N+D
        more = 1000*M+100*O+10*R+E
        money = 10000*M+1000*O+100*N+10*E+Y
        if send+more == money:
            print(send, "+", more, "=", money)
            break
```

1.5 Missionaries and Cannibals

```python
from collections import deque

q = deque([((3,3,1), [])])
seen = set()
moves = [(1,0),(2,0),(0,1),(0,2),(1,1)]

while q:
    (m,c,b), path = q.popleft()
    if (m,c,b) == (0,0,0):
        print(path+[(m,c,b)])
        break
    if (m,c,b) in seen:
        continue
    seen.add((m,c,b))
    for x,y in moves:
        nm,nc = (m-x,c-y) if b else (m+x,c+y)
        if 0<=nm<=3 and 0<=nc<=3 and (nm==0 or nm>=nc) and (3-nm==0 or 3-nm>=3-nc):
            q.append(((nm,nc,1-b),path+[(m,c,b)]))
```

1.7 Breadth First Search (BFS)

```python
from collections import deque

g = {"A":["B","C"], "B":["D","E"],
     "C":["F"], "D":[], "E":[], "F":[]}
q = deque(["A"])
seen = {"A"}

while q:
    n = q.popleft()
    print(n, end=" ")
    for x in g[n]:
        if x not in seen:
            seen.add(x)
            q.append(x)
```

1.8 Depth First Search (DFS)

```python
g = {"A":["B","C"], "B":["D","E"],
     "C":["F"], "D":[], "E":[], "F":[]}
seen = set()

def dfs(n):
    if n not in seen:
        print(n, end=" ")
        seen.add(n)
        for x in g[n]:
            dfs(x)

dfs("A")
```

1.9 Travelling Salesman Problem

```python
from itertools import permutations

g = [[0,10,15,20],
     [10,0,35,25],
     [15,35,0,30],
     [20,25,30,0]]

best = float("inf")
for p in permutations([1,2,3]):
    path = (0,) + p + (0,)
    cost = sum(g[path[i]][path[i+1]] for i in range(4))
    if cost < best:
        best, route = cost, path

print("Path:", route)
print("Cost:", best)
```

1.10 A* Algorithm

```python
import heapq

g = {"A":[("B",1),("C",3)],
     "B":[("D",3),("E",1)], "C":[("F",5)],
     "D":[("G",2)], "E":[("G",5)],
     "F":[("G",2)], "G":[]}
h = {"A":7,"B":6,"C":5,"D":2,"E":4,"F":2,"G":0}

q = [(h["A"], 0, "A", ["A"])]
while q:
    f, cost, n, path = heapq.heappop(q)
    if n == "G":
        print("Path:", path, "Cost:", cost)
        break
    for x, w in g[n]:
        heapq.heappush(q, (cost+w+h[x], cost+w, x, path+[x]))
```
